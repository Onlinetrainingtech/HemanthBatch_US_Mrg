```python
from scipy.cluster.hierarchy import dendrogram,linkage
```

```python
from matplotlib import pyplot as plt
%matplotlib inline
```

```python
import numpy as np
import pandas as pd
```

```python
plt.rcParams['figure.figsize']=(5,5)
plt.style.use('ggplot')
```

```python
data=pd.read_csv('Datasets/movie_metadata1.csv')
f1=data['budget'].values
f2=data['gross'].values
```

```python
fb=f1[0:10]
```

```python
fg=f2[0:10]
```

```python
X=np.array(list(zip(fb,fg)))
Z=linkage(X,'ward')
```

```python
fig=plt.figure(figsize=(5,5))
dn=dendrogram(Z)
Z=linkage(X,'single')
```

```python
fig=plt.figure(figsize=(5,5))
dn=dendrogram(Z)
plt.show()
```
