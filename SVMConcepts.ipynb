```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn import datasets
```

```python
iris=datasets.load_iris()
```

```python
print(iris.DESCR)
```

```python
from sklearn import svm
X=iris.data[:,:2]
Y=iris.target
```

```python
#SVM Configuration parameters
C=1.0
svc=svm.SVC(kernel='linear',C=C).fit(X,Y)
```

```python
svc
```

```python
rbf_svc=svm.SVC(kernel='rbf',gamma=0.5,C=C).fit(X,Y)
```

```python
rbf_svc
```

```python
poly_svm=svm.SVC(kernel='poly',degree=3,C=C).fit(X,Y)
poly_svm
```

```python
#Spliting the data
from sklearn.model_selection import train_test_split
X_train,X_test,Y_train,Y_test=train_test_split(X,Y,test_size=0.3)
```

```python
#SVC linear
lin_svc=svm.LinearSVC(C=C).fit(X,Y)
```

```python
h=0.02 #step size
x_min=X[:,0].min()-1
x_max=X[:,0].max()+1
y_min=X[:,1].min()-1
y_max=X[:,1].max()+1
xx,yy=np.meshgrid(np.arange(x_min,x_max,h),np.arange(y_min,y_max,h))
yy
```

```python
arr=np.arange(x_min,x_max,h)
arr
```

```python
arr1=np.arange(y_min,y_max,h)
arr1
```

```python
titles=['SVC with linear kernel','LinearSVC(Linear kernel)','SVC with RBF kernel','SVC with polynomial(degree 3)']
```

```python
for i,clf in enumerate((svc,lin_svc,rbf_svc,poly_svm)):
    plt.figure(figsize=(15,15))
    plt.subplot(2,2,i+1)
    plt.subplots_adjust(wspace=0.4,hspace=0.4)
    Z=clf.predict(np.c_[xx.ravel(),yy.ravel()])
    Z=Z.reshape(xx.shape)
    plt.contourf(xx,yy,Z,cmap=plt.cm.terrain,alpha=1)
    plt.scatter(X[:,0],X[:,1],c=Y,cmap=plt.cm.Dark2)
    plt.xlabel('Sepal length')
    plt.ylabel('Sepal width')
    plt.xlim(xx.min(),xx.max())
    plt.ylim(yy.min(),yy.max())
    plt.xticks()
    plt.yticks()
    plt.title(titles[i])
```
