```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
from sklearn.model_selection import train_test_split
from sklearn import metrics
from sklearn import tree
```

```python
data=pd.read_csv('Datasets/pacific.csv')
data.head()
```

```python
#2.DataWrangling
```

```python
pred_columns=data[:]
```

```python
pred_columns.drop(['ID','Name','Date','Event','Status','Latitude','Longitude'],inplace=True,axis=1)
pred_columns.head()
```

```python
#list of input columns
input_columns=list(pred_columns)
input_columns
```

```python
sns.countplot(data['Status'],label='count')
plt.show()
```

```python
#Spliting the data into training and testing
train,test=train_test_split(data,test_size=0.3)
```

```python
train_X=train[input_columns]
train_Y=train['Status']
```

```python
test_X=test[input_columns]
test_Y=test['Status']
```

```python
#Model
model=tree.DecisionTreeClassifier()
```

```python
#Training the model
model.fit(train_X,train_Y)
predicition=model.predict(test_X)
```

```python
metrics.accuracy_score(predicition,test_Y)
```

```python
d=pd.DataFrame({'predicated_value':predicition,'KnownO/P':test_Y})
d
```

```python
#http://www.webgraphviz.com/
tree.export_graphviz(model,out_file='decision_tree.dot',feature_names=input_columns)
```
